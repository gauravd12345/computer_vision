```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dansbecker/food-101")

print("Path to dataset files:", path)
```

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

base_path = Path(path)

images = base_path / "food-101" / "food-101" / "images"
foods = sorted([p.name for p in images.iterdir() if p.is_dir()])

fig, axes = plt.subplots(1, 10, figsize=(20, 3))

for ax, food in zip(axes, foods[:10]):
    img_path = next((images / food).iterdir())

    with Image.open(img_path) as img:
        img = img.resize((100, 100))

        ax.imshow(img)
        ax.set_title(food, fontsize=8)
        ax.axis("off")

plt.tight_layout()
plt.show()
```
